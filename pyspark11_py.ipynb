```python

from pyspark.sql import SparkSession
# Start Spark
spark = SparkSession.builder \
    .appName("Student") \
    .getOrCreate()
# Create student data
data = [
    ("Alemu", 85),
    ("Beti", 72),
    ("Selam", 90),
    ("Dawit", 65),
    ("Eleni", 95)
]
columns = ["Name", "Score"]
df = spark.createDataFrame(data, columns)
# Show all students
df.show()
# Calculate average score
df.groupBy().avg("Score").show()
# Show students with score > 80
df.filter(df.Score > 80).show()
```
